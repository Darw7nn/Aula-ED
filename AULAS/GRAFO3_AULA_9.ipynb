```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()

G.add_nodes_from(['Elvis', 'Fã-1', 'Fã-2','Fã-3','Não-Fã'])
```

```python
G.add_edges_from([('Fã-1', 'Elvis'),
                  ('Fã-2', 'Elvis'),
                  ('Fã-3', 'Elvis'),])
```

```python
print(f"Nós do grafo: {G.nodes()}")
print(f"Arestas do grafro: {G.edges()}")
```

```python
pos = nx.spring_layout(G, seed=42)

nx.draw(G, pos, with_labels= True, node_color='red',
        font_weight='bold',node_size=1000, font_size=10)

plt.title('Representação em Grafo simples')
plt.show()
```
