```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()

G.add_nodes_from(['Beyonce', 'David Bowie', 'Jay-Z','Kanye West','Lady Gaga','Prince', 'Madonna','Michael Jackson'])
```

```python
G.add_edges_from([('Beyonce', 'Rihanna'),
                  ('David Bowie', 'Prince'),
                  ('David Bowie', 'Madonna'),
                  ('Jay-Z', 'Rihanna'),
                  ('Jay-Z', 'Kanye West'),
                  ('Kanye West', 'Rihanna'),
                  ('Madonna','Lady Gaga'),
                  ('Michael Jackson', 'Beyonce'),
                  ('Prince', 'Michael Jackson'),
                  ('Michael Jackson','Beyonce'),])
```

```python
print(f"Nós do grafo: {G.nodes()}")
print(f"Arestas do grafro: {G.edges()}")
```

```python
pos = nx.spring_layout(G, seed=42)

nx.draw(G, pos, with_labels= True, node_color='pink',
        font_weight='bold',node_size=800, font_size=7)

plt.title('Representação em Grafo simples')
plt.show()
```
