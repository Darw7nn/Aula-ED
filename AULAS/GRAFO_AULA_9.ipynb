```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()

G.add_nodes_from(['Harry', 'Gayle', 'Frank', 'Emily', 'Dave', 'Cathy', 'Bill', 'Jeff', 'Audrey', 'Llana'])

```

```python
G.add_edges_from([
    ('Gayle', 'Bill'),
    ('Gayle', 'Audrey'),
    ('Gayle', 'Cathy'),
    ('Frank', 'Emily'),
    ('Frank', 'Dave'),
    ('Audrey', 'Bill'),
    ('Audrey', 'Llana'),
    ('Llana', 'Harry'),
    ('Harry', 'Jeff'),
    ('Emily', 'Jeff'),
    ('Llana', 'Dave'),
    ('Cathy', 'Emily'),
    ('Bill', 'Emily'),
    ('Bill', 'Jeff'),
    ('Emily', 'Dave'),])
```

```python
print(f"Nós do grafo: {G.nodes()}")
print(f"Arestas do grafro: {G.edges()}")
print(f"Contando o num de Nós: {G.number_of_nodes()}")
print(f"Contando o num de Arestas: {G.number_of_edges()}")
```

```python
pos = nx.spring_layout(G, seed=42)

nx.draw(G, pos, with_labels= True, node_color='lightblue',
        font_weight='bold',node_size=700, font_size=10)

plt.title('Representação em Grafo simples')
plt.show()
```
